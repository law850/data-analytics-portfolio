```python
# Merging DataFrames:
# Create two DataFrames:
#df1 = pd.DataFrame({
#    'ID': [1, 2, 3],
 #   'Name': ['Alice', 'Bob', 'Charlie']
#})
#df2 = pd.DataFrame({
#   'ID': [2, 3, 4],
#'Salary': [70000, 80000, 90000]
#})
#• Merge the two DataFrames on the ID column using an inner join.
import pandas as pd

# 1. Create the first DataFrame (df1)
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

# 2. Create the second DataFrame (df2)
df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Salary': [70000, 80000, 90000]
})

# 3. Merge the two DataFrames on the 'ID' column using an inner join
merged_df = pd.merge(df1, df2, on='ID', how='inner')

# 4. Display the result
print("DataFrame 1:")
display(df1)
print("\nDataFrame 2:")
display(df2)
print("\nMerged DataFrame (Inner Join):")
display(merged_df)
```

```python
import pandas as pd

# 1. Create the "Customers" table (from the CRM system)
customers_df = pd.DataFrame({
    'CustomerID': [101, 102, 103, 104],
    'CustomerName': ['Alice Smith', 'Bob Jones', 'Charlie Brown', 'Diana Prince'],
    'Email': ['alice@email.com', 'bob@email.com', 'charlie@email.com', 'diana@email.com']
})

# 2. Create the "Orders" table (from the Sales system)
orders_df = pd.DataFrame({
    'OrderID': ['ORD-001', 'ORD-002', 'ORD-003', 'ORD-004'],
    'CustomerID': [102, 103, 105, 101], # Notice 104 didn't order, and 105 is a guest checkout
    'OrderAmount': [150.00, 200.50, 75.25, 300.00]
})

# 3. Perform the Inner Join
# We merge on 'CustomerID' because that is the common column (the key) in both tables.
customer_orders = pd.merge(customers_df, orders_df, on='CustomerID', how='inner')

# 4. Display the results
print("--- Customers Table ---")
display(customers_df)
print("\n--- Orders Table ---")
display(orders_df)
print("\n--- Merged Table (Inner Join) ---")
display(customer_orders)
```

```python
import pandas as pd

# 1. Create the "Employees" table (from the HR database)
employees_df = pd.DataFrame({
    'EmpID': [201, 202, 203, 204, 205],
    'Name': ['Sarah', 'Mike', 'Jessica', 'David', 'Emily'],
    'Department': ['Sales', 'Engineering', 'Sales', 'Marketing', 'Engineering']
})

# 2. Create the "Performance" table (from the quarterly review system)
performance_df = pd.DataFrame({
    'EmpID': [202, 203, 205, 206], # 201 and 204 are missing (maybe they are new hires)
    'Performance_Score': [85, 92, 78, 88], # 206 is a contractor not in the HR table
    'Review_Date': ['2023-10-01', '2023-10-02', '2023-10-01', '2023-10-03']
})

# 3. Perform the Inner Join
# We merge on 'EmpID' because it is the unique identifier in both tables.
performance_review = pd.merge(employees_df, performance_df, on='EmpID', how='inner')

# 4. Display the results
print("--- Employees Table ---")
display(employees_df)
print("\n--- Performance Table ---")
display(performance_df)
print("\n--- Merged Table (Inner Join) ---")
display(performance_review)
```

```python
import pandas as pd

# 1. Create the "Products" table
products_df = pd.DataFrame({
    'ProductID': ['P01', 'P02', 'P03', 'P04', 'P05'],
    'ProductName': ['Laptop', 'Mouse', 'Monitor', 'Keyboard', 'Webcam'],
    'Category': ['Electronics', 'Accessories', 'Electronics', 'Accessories', 'Accessories'],
    'UnitPrice': [1200.00, 25.00, 300.00, 75.00, 50.00]
})

# 2. Create the "Transactions" table
transactions_df = pd.DataFrame({
    'TransactionID': ['T001', 'T002', 'T003', 'T004', 'T005', 'T006'],
    'ProductID': ['P01', 'P02', 'P01', 'P03', 'P05', 'P06'], # P06 is a new product not in our catalog
    'Quantity': [2, 5, 1, 3, 4, 2]
})

# 3. Merge the two tables on 'ProductID' using an inner join
sales_df = pd.merge(transactions_df, products_df, on='ProductID', how='inner')

# 4. Create a new column for Total Revenue
sales_df['Total_Revenue'] = sales_df['Quantity'] * sales_df['UnitPrice']

# 5. Display the merged table
print("--- Merged Sales Table ---")
display(sales_df)

# 6. Group by Category and sum the revenue
category_revenue = sales_df.groupby('Category')['Total_Revenue'].sum().reset_index()

print("\n--- Total Revenue by Category ---")
display(category_revenue)
```
