```python
import pandas as pd
#1. DataFrame Creation:
#o Create a Pandas DataFrame from the following dictionary:
#data = {
#      'Name': ['Alice', 'Bob', 'Charlie', 'David'],
#       'Age': [25, 30, 35, 40],
#       'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
#       }
#      o Display the DataFrame.

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}

df = pd.DataFrame(data)
display(df)
```

```python
import numpy as np

# Example 1: Creating a DataFrame from a list of dictionaries
data_dicts = [
    {'Product': 'Apples', 'Price': 1.20, 'InStock': True},
    {'Product': 'Bananas', 'Price': 0.50, 'InStock': True},
    {'Product': 'Cherries', 'Price': 3.00, 'InStock': False}
]
df_dicts = pd.DataFrame(data_dicts)
print("--- DataFrame from List of Dictionaries ---")
display(df_dicts)

# Example 2: Creating a DataFrame from a list of lists
data_lists = [
    ['Alice', 'Engineering', 85000],
    ['Bob', 'Marketing', 65000],
    ['Charlie', 'Sales', 70000]
]
columns = ['Employee', 'Department', 'Salary']
df_lists = pd.DataFrame(data_lists, columns=columns)
print("\n--- DataFrame from List of Lists ---")
display(df_lists)
```
